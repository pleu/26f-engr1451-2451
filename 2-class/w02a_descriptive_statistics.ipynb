## 1. Setup and load the Iris data


```python
from pathlib import Path

import numpy as np
import pandas as pd
from IPython.display import display

data_path = Path("./data/iris.csv")
iris = pd.read_csv(data_path)

iris.head()
```

## 2. Inspect rows, columns, and variable types

Start with structure before calculating statistics. A row is one observation, a column is one variable, and each column has a stored data type.


```python
print(iris.dtypes)

print("Missing values:")
print(iris.isna().sum())

print(iris["variety"].value_counts())
```

## 3. Select variables and observations

- `df["column"]` selects one column as a pandas `Series`.
- `df[["a", "b"]]` selects several columns as a DataFrame.
- `.loc[condition, columns]` filters rows and chooses columns by name.
- Use `:` to indicate all


```python
petal_length = iris["petal.length"]
petal_measurements = iris[["petal.length", "petal.width"]]

print("One column becomes a", type(petal_length).__name__)
print(petal_length.head())

print("\nTwo columns remain a", type(petal_measurements).__name__)
display(petal_measurements.head())
```

```python
versicolor = iris.loc[
    iris["variety"] == "Versicolor",
    ["sepal.length", "sepal.width", "petal.length", "petal.width", "variety"],
]

display(versicolor.head())

versicolor_all = iris.loc[
  iris["variety"] == "Versicolor",
  :
]
display(versicolor_all.head())
```

## 4. Measures of center

Use one quantitative Iris variable as a pandas `Series`. The mean and median are single values; `.mode()` can return more than one value when several measurements tie for highest frequency.

`agg()` means aggregate. It applies several summary functions to the same Series.


```python
petal_length = iris["petal.length"]

print(petal_length.count())
print(petal_length.mean())
print(petal_length.median())

center_summary = petal_length.agg(["count", "mean", "median"])
print(center_summary)

print("Mode(s):", petal_length.mode())
```

### Translate the sample-mean equation into code

The slide equation is

$$
\bar{x}=\frac{\sum_{i=1}^{n}x_i}{n}.
$$

The direct calculation and the pandas method should agree.


```python
petal_length = iris["petal.length"]

mean_from_equation = petal_length.sum() / petal_length.count()
mean_from_pandas = petal_length.mean()

print("sum / n:", mean_from_equation)
print("Series.mean():", mean_from_pandas)
```

## 5. Measures of spread

Now use `petal.length` from the Iris dataset. The range uses only the minimum and maximum. Variance and standard deviation use every deviation from the mean.


```python
petal_length = iris["petal.length"]
mean_value = petal_length.mean()

print("Range:", petal_length.max() - petal_length.min())
```

### Sample versus population variance

pandas uses the sample convention by default:

- `s.var()` and `s.std()` divide by $n-1$;
- `s.var(ddof=0)` and `s.std(ddof=0)` divide by $n$.


```python
petal_length = iris["petal.length"]

n = petal_length.count()
sum_squared_deviations = ((petal_length - petal_length.mean()) ** 2).sum()

print("sample variance", petal_length.var())
print("population variance", petal_length.var(ddof=0))
print("sample standard deviation", petal_length.std())
print("population standard deviation", petal_length.std(ddof=0))
```

## 6. Overall and grouped summaries

An overall mean can hide meaningful subgroups. Compare pooled and species-specific summaries before interpreting the dataset.

Use `describe()` to calculate standard summary statistics


```python
numerical_columns = ["sepal.length", "sepal.width", "petal.length", "petal.width"]

overall_summary = iris[numerical_columns].describe()
overall_summary
```

```python
petal_length_by_species = (
    iris.groupby("variety")["petal.length"]
    .agg(["count", "mean", "median", "std", "min", "max"])
)

petal_length_by_species
```

```python
pooled_and_group_means = pd.concat(
    [
        pd.Series({"all species combined": iris["petal.length"].mean()}),
        iris.groupby("variety")["petal.length"].mean(),
    ]
).to_frame("mean petal.length")

pooled_and_group_means
```

## 7. Compact code reference

| Goal | Python pattern |
|---|---|
| First rows | `df.head()` |
| Dimensions | `df.shape` |
| Data types | `df.dtypes` |
| Missing-value counts | `df.isna().sum()` |
| One variable | `df["column"]` |
| Filter rows | `df.loc[condition, columns]` |
| Mean / median / mode | `s.mean()`, `s.median()`, `s.mode()` |
| Range | `s.max() - s.min()` |
| Sample variance / SD | `s.var()`, `s.std()` |
| Population variance / SD | `s.var(ddof=0)`, `s.std(ddof=0)` |
| Fast summary | `df.describe()` |
| Summaries by group | `df.groupby("group")["value"].agg(...)` |
