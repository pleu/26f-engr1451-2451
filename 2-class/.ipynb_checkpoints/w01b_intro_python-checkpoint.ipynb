## 1. Jupyter Notebook Essentials

A Jupyter notebook combines two main cell types:

- **Markdown cells** contain headings, explanations, equations, links, and instructions.
- **Code cells** contain Python code and display results directly below the cell.

Useful commands:

- `Shift` + `Enter`: run the current cell and move to the next cell.
- `Ctrl` + `Enter`: run the current cell and stay in it.
- Double-click a Markdown cell to edit it.
- Press `Esc` to enter command mode and `Enter` to return to edit mode.

A notebook remembers variables created by previously executed cells. Therefore, **execution order matters**. Before submitting a notebook, restart the kernel and run all cells from top to bottom.

## 2. Python Basics

A **variable** gives a name to a value. Python determines the data type from the assigned value; no separate type declaration is needed.


````python
# Anything after a # is a comment
course = "Exploratory Data Science"  # str
week = 1                             # int
temperature_c = 22.5                 # float
is_ready = True                      # bool

print(course)
print(week, temperature_c, is_ready)
````

### Common data types

| Type | Example | Meaning |
|---|---:|---|
| `int` | `7` | Integer |
| `float` | `7.5` | Decimal number |
| `str` | `"Pittsburgh"` | Text |
| `bool` | `True` | Logical value |

Use `type()` to inspect a value.


````python
type(course), type(week), type(temperature_c), type(is_ready)
````

### Names, indentation, and errors

- Python names are **case-sensitive**: `temperature` and `Temperature` are different names.
- Use descriptive names such as `average_temperature`, not vague names such as `x1`.
- Indentation defines code blocks for `if` statements, loops, and functions.
- When an error occurs, read the **last line of the error message first**. It usually identifies both the error type and the immediate problem.

Common errors include `NameError`, `TypeError`, `SyntaxError`, and `IndentationError`.

## 3. Operators

Python provides arithmetic, comparison, and Boolean operators.


````python
a = 7
b = 3

print("addition:", a + b)
print("subtraction:", a - b)
print("multiplication:", a * b)
print("division:", a / b)
print("integer division:", a // b)
print("remainder:", a % b)
print("power:", a ** b)
````

Important arithmetic operators:

- `/` performs ordinary division.
- `//` performs floor (integer) division.
- `%` returns the remainder.
- `**` raises a number to a power.


````python
temperature_c = 22.5

print(temperature_c > 20)
print(temperature_c == 22.5)
print(20 <= temperature_c < 25)
print((temperature_c > 20) and is_ready)
print((temperature_c < 0) or (temperature_c > 35))
print(not is_ready)
````

Comparison operators include `==`, `!=`, `<`, `<=`, `>`, and `>=`.

Boolean expressions can be combined with `and`, `or`, and `not`. Remember that `=` assigns a value, whereas `==` checks whether two values are equal.
